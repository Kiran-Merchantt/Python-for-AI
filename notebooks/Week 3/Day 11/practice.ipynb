Build a shape family. `Shape` is the base with an `area()` that returns `0`. `Rectangle(width, height)` and `Circle(radius)` each override `area()`. Put a few in a list and print each area in one loop (polymorphism).


```python
class shape:
    def area(self):
        return 0


class rectangle(shape): 
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

class circle(shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2


shapes = [
    rectangle(3, 5),
    circle(5)
]

for shape in shapes:
    print(shape.area())
        
```
