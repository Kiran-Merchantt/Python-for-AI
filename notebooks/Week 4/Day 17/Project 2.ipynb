```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

try:
    df = pd.read_csv("tips.csv")
except FileNotFoundError:
    sns.load_dataset("tips").to_csv("tips.csv", index=False)   # needs internet
    df = pd.read_csv("tips.csv")
```
