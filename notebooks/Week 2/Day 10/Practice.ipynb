Write a `Student` class. `__init__` takes a `name` and a list of `marks`. Add a method `average()` that returns the mean mark, and a method `grade()` that returns `"A"` for 80+, `"B"` for 70+, `"C"` for 60+, else `"F"`. Test it on `Student("Fatima", [85, 92, 78])`.


```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def average(self):
        return sum(self.marks) / len(self.marks)

    def grade(self):
        avg = self.average()
        if avg >= 80:
            return "A"
        elif avg  >= 70:
            return "B"
        elif avg >= 60:
            return "C"
        else:
            "D"

Fatima = Student("Fatima", [85, 92, 78])

print(Fatima.average())
print(Fatima.grade())



    
```

Add a `transfer(self, other, amount)` method to `BankAccount` that moves money from this account to `other`, but only if this account has enough. Reuse the `withdraw`/`deposit` methods you already wrote — don't touch the balances directly.

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        self.balance -= amount

    def transfer(self, other, amount):
        if(self.balance >= amount):
            self.withdraw(amount)
            other.deposit(amount)
        else:
            print("You dont have suffient funds")

ali = BankAccount("Ali", 5000)
zara = BankAccount("Zara", 5000)

ali.transfer(zara, 6000)

print("Balance of Ali:", ali.balance)
print("Balance of Zara:", zara.balance)
```
