```python
students = [("Ali", 85), ("Fatima", 92), ("Hassan", 78)]

students.sort(key=lambda student: student[1])   # sort by the second item (marks)
print(students)


# students = [("Ali", 85), ("Fatima", 92), ("Hassan", 78)]

# students.sort(key=lambda student: student[1], reverse=True)   # sort by the second item (marks)
# print(students)
```

```python
# Given `[("Ali", 22), ("Zara", 19), ("Hamza", 25)]`, sort it by age using `.sort()` and a `lambda`, then print it.

students = [("Ali", 22), ("Zara", 19), ("Hamza", 25)]

students.sort(key=lambda age: age[1])

print(students)
```
