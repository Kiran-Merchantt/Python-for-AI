```python
def countdown(n):
    if n == 0:
        print("Done")
        return
    print(n)
    countdown(n - 1)

countdown(5)
```

```python
def countdown(n):
    if n == 0:
        return 0
    return n + countdown(n-1)

print(countdown(3))
```

```python
def reverse(text):
    if text == "":
        return ""
    return reverse(text[1:]) + text[0]
```

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

```python
def fac(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

print(fac(5))
```

```python
expenses = {
    "rent": 80000,
    "salaries": {
        "cashiers": 120000,
        "managers": 90000,
    },
    "marketing":{
        "online": {
            "facebookads": 25000,
            "googleads" : 15000,
        },
        "flyers": 5000,
    },
        "utilities": 18000,
}

def totalcost(item):
    if type(item) != dict:
        return item
    amount = 0
    for child in item.values():
        return amount

revenue = 500000
total_expenses = totalcost(expenses) 
print("Revenue:       ", revenue)
print("Total expenses:", total_expenses)
print("Profit:        ", revenue - total_expenses)
print()
print("Marketing alone:", totalcost(expenses["marketing"]))
```
